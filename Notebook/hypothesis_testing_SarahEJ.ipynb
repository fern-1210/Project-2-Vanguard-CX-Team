```python
import pandas as pd
import scipy.stats as st
import numpy as np

df = pd.read_csv(r"C:\Users\User\Documents\Iron Hack\Week5\Project\Project-2-Vanguard-CX-Team\Data\Clean Data\analysis.csv")
```

```python
#Define the hypotheses

#H0: average client tenure is the same for both groups/ mean of clnt_tenure_yr_control = mean of clnt_tenure_yr_test
#H1: average client tenure is different for the two groups/mean of clnt_tenure_yr_control != mean of clnt_tenure_yr_test

#choose the significance level: 0.05

#obtain average tenures and variance for both groups

stats = df.groupby('Variation')['clnt_tenure_yr'].agg(['mean', 'var']).reset_index()

stats
```

```python
from scipy.stats import ttest_ind

# Extract the tenure data for each group
tenure_test = df[df['Variation'] == 'Test']['clnt_tenure_yr']
tenure_control = df[df['Variation'] == 'Control']['clnt_tenure_yr']

# Perform a two-sample t-test
t_stat, p_value = ttest_ind(tenure_test, tenure_control, equal_var=False)  # Welch's t-test

# Print the results
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Decision
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis. There is a statistically significant difference in average client tenure.")
else:
    print("Fail to reject the null hypothesis. There is no statistically significant difference in average client tenure.")
```
