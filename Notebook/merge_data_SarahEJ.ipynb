```python
#Who are the primary clients using this online process? primary= highest balance

#Identify Primary Clients: First, filter clients based on the highest balances. Let's assume we want the top 10%:
balance_threshold = merged_df['bal'].quantile(0.99)  # Top 1% threshold
primary_clients = merged_df[merged_df['bal'] >= balance_threshold]

primary_clients
```

```python
# sort by balance in descending order for clarity
primary_clients = primary_clients.sort_values(by='bal', ascending=False)

# Drop duplicates based on 'client_id' to ensure unique client IDs
primary_clients = primary_clients.drop_duplicates(subset='client_id')

primary_clients
```

```python
primary_clients[['clnt_age', 'clnt_tenure_yr', 'gendr','bal', 'num_accts']].describe()
```

```python
# Calculate median for each numeric column
median_age = primary_clients['clnt_age'].median()
median_balance = primary_clients['bal'].median()
median_tenure = primary_clients['clnt_tenure_yr'].median()

# Calculate mode for each numeric column
mode_age = primary_clients['clnt_age'].mode()
mode_balance = primary_clients['bal'].mode()
mode_tenure = primary_clients['clnt_tenure_yr'].mode()

# Print outcomes
print(f"Median Age: {median_age}")
print(f"Median Balance: {median_balance}")
print(f"Median Tenure: {median_tenure}")

print("\nMode Age:")
print(mode_age)
print("\nMode Balance:")
print(mode_balance)
print("\nMode Tenure:")
print(mode_tenure)
```
